```python
def get_list_of_sentences(file_path):
    """this returns a list of lists for sentences and list of lists for labels"""
    current_sent = []
    current_label = []
    all_sents = []
    all_labels = []

    with open(file_path, "r", encoding="utf-8") as infile:
        lines = infile.readlines()
        for line in lines:
            line = line.strip()
            if line == "":
                if len(current_sent) > 0:
                    all_sents.append(current_sent)
                    all_labels.append(current_label)
                    current_sent = []
                    current_label = []
            else:
                splitted = line.split("\t")
                token_part = splitted[0]
                label_part = splitted[1]
                current_sent.append(token_part)
                current_label.append(label_part)

        if len(current_sent) > 0:
            all_sents.append(current_sent)
            all_labels.append(current_label)
    
    return all_sents, all_labels
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from pathlib import Path
from matplotlib.colors import LogNorm
from collections import defaultdict, Counter
```

```python
def nervaluate_macro(evaluator, scheme="strict"):
    """this calculates macro-averaged precision, recall, and F1."""
    results = evaluator.evaluate()["entities"]
    precisions = []
    recalls = []
    f1_scores = []
    for label in results:
        scores = results[label][scheme]
        precisions.append(scores.precision)
        recalls.append(scores.recall)
        f1_scores.append(scores.f1)
    return {"precision": sum(precisions) / len(precisions), "recall": sum(recalls) / len(recalls), "f1": sum(f1_scores) / len(f1_scores)}
```

```python
def plot_confusion_matrix(cm, title, outpath):
    """this plots and saves a cf matrix -- claude was used to generate the confusion matrix"""

    width = max(6, 0.6 * len(cm.columns))
    height = max(5, 0.5 * len(cm.index))
    fig, ax = plt.subplots(figsize=(width, height))
    sns.heatmap(cm, annot=False, cmap="BuPu", cbar=True, linewidths=0.5, linecolor="white", norm=LogNorm(vmin=1, vmax=cm.values.max()), ax=ax)
    threshold = cm.values.max() ** 0.5
    for row in range(cm.shape[0]):
        for column in range(cm.shape[1]):
            value = cm.iloc[row, column]
            if value > threshold:
                text_color = "white"
            else:
                text_color = "black"
            ax.text(column + 0.5, row + 0.5, str(value), ha="center", va="center", color=text_color, fontsize=9)

    ax.set_xlabel("Predicted")
    ax.set_ylabel("Gold")
    ax.set_title(title)
    plt.xticks(rotation=45, ha="right")
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.savefig(outpath, dpi=150, bbox_inches="tight")
    plt.close(fig)


def strip_prefix(tag):
    """removes bio labeling."""
    if tag == "O":
        return "O"
    return tag[2:]

gold_token = []
for sentence in test_labels:
    for tag in sentence:
        gold_token.append(strip_prefix(tag))
pred_token = []
for sentence in pred_labels:
    for tag in sentence:
        pred_token.append(strip_prefix(tag))
token_cm = pd.crosstab(pd.Series(gold_token, name="gold"), pd.Series(pred_token, name="pred"), dropna=False)
print("Token-level confusion matrix:")
print(token_cm.to_string())
plot_confusion_matrix(token_cm, title="Entity Type-level Confusion Matrix for GLiNER-LLM", outpath="GLINER_LLM_confusion_token.png"
)
```

```python
def find_misclassifications(test_sents, test_labels, pred_labels, context_window=3):
    """this finds tokens where the pred label is different from the gold label
    -- claude used for trouble shooting where needed
    """
    errors = defaultdict(list)
    for sentence_index in range(len(test_sents)):  #numbering every sentence
        tokens = test_sents[sentence_index]
        gold_labels = test_labels[sentence_index] #getting the gold labels per sentence
        predicted_labels = pred_labels[sentence_index] #getting the pred labels per sentence
        for token_index in range(len(tokens)): #for tok index in all tokens in a sentence
            gold_label = gold_labels[token_index] # get each gold label
            predicted_label = predicted_labels[token_index] #and pred label
            gold_type = strip_prefix(gold_label) #and we strip the bio label!
            pred_type = strip_prefix(predicted_label) #for this pred label too we just need the type
            if gold_type == pred_type: # if the same, ignore since we only want misclassifications
                continue
            start = max(0, token_index - context_window) #otherwise, find the starting of the context window
            end = min(len(tokens), token_index + context_window + 1) # and get the ending of the context window
            context = tokens[start:end] #now we have the context window to analyze too!
            target_position = token_index - start #now we also get the starting position of the token
            errors[(gold_type, pred_type)].append({"sentence_index": sentence_index, "token_index": token_index, "token": tokens[token_index], "context": context, "target_position": target_position, "gold_label": gold_label, "pred_label": predicted_label})
    return errors # we have our list of error info per gold label/pred label misclassified pair!


def print_misclassifications(errors, top_n_per_pair=5, min_count=2):
    """this prints the most common misclassification types w examples"""
    sorted_errors = sorted(errors.items(), key=lambda item: len(item[1]), reverse=True) #lets sort the error types from most to least frequent
    for (gold_type, pred_type), examples in sorted_errors: #for every gold label, pred label misclassified pair in the sorted errors
        count = len(examples) #we count the # of examples
        if count < min_count: #if less than the minimum count, we continue since those are pretty rare
            continue
        print(f"\n{'=' * 70}")
        print(f"  GOLD: {gold_type:<15s}  PRED: {pred_type:<15s}  (count: {count})")
        print("=" * 70)
        token_counts = Counter()
        for example in examples: #for every example in the misclassified examples
            token = example["token"] #we get the token text
            token_counts[token] += 1 #and keep a counter for it!
        print("most frequent misclassified tokens:")
        for token, count in token_counts.most_common(10): #now we get the top 10 tokens based on most frequent errors
            print(f"    '{token}' ({count}x)")
        number_of_examples = min(top_n_per_pair, count) #returning the example quantity we want
        print(f"\nexamples (showing {number_of_examples}):")
        for example in examples[:top_n_per_pair]: #for examples in each pair
            context = example["context"][:] #making a copy of the context for marking the error
            target_position = example["target_position"] #the error's position
            context[target_position] = (f"**{context[target_position]}**") #so we can get the correct context window and mark the error for clarity
            print( #print them for analysis
                f"    [sent {example['sentence_index']}, "
                f"tok {example['token_index']}] "
                f"gold={example['gold_label']}, "
                f"pred={example['pred_label']}"
            )
            print(f"      ...{' '.join(context)}...")

errors = find_misclassifications(test_sents, test_labels, pred_labels, context_window=15)
print_misclassifications(errors, top_n_per_pair=30, min_count=3)
```

### SPAN LEVEL

```python
##  -- claude used for trouble shooting where needed
def bio_to_spans(gold_labels):
    """this converts BIO tags into entity spans
    """
    spans = []
    current_label = None #NOT inside an entity
    start_index = None
    for index, gold_label in enumerate(gold_labels): #for num, gold label in all gold labels
        if gold_label == "O": #if the gold label is "O", not inside an entity!
            if current_label is not None: #if inside an entity it needs to be closed
                spans.append((current_label, start_index, index)) #we retain the current label, its start index amd end index
                current_label = None # then reset back to none since the entity is closing
                start_index = None #reset this too
            continue
        prefix, dash, label = gold_label.partition("-") #separate the BIO tags from the label types
        if prefix == "B" or label != current_label: #if its B- or the label differs from the last stored current label, its a new entity
            if current_label is not None: # close the last entity being tracked
                spans.append((current_label, start_index, index)) #save the previous entity's label, its starting index and end index
            current_label = label #now we set it to the entity type
            start_index = index #and set a starting index
    if current_label is not None: #if in an ongoing entity, add its label, start index and length
        spans.append((current_label, start_index, len(gold_labels))) #end the ongoing entity at the end of the sentence
    return spans

def find_boundary_mismatches(test_sents, test_labels, pred_labels):
    """this finds entities with the correct type but incorrect boundaries."""
    results = []
    for sentence_index in range(len(test_sents)): #for sentence in test sents,
        tokens = test_sents[sentence_index] # we get the tokens
        gold_labels = test_labels[sentence_index] #and the gold labels
        predicted_labels = pred_labels[sentence_index] #and predicted labels
        gold_spans = bio_to_spans(gold_labels) #and then their gold spans
        predicted_spans = bio_to_spans(predicted_labels) #and their predicted spans
        for gold_label, gold_start, gold_end in gold_spans: #for every gold label, its start index and end index in gold spans
            for predicted_label, predicted_start, predicted_end in predicted_spans: #we match it to every predicted label, with its start index and end index in pred spans
                if gold_label != predicted_label: #skip ones that dont have the same label
                    continue
                spans_overlap = (gold_start < predicted_end and predicted_start < gold_end) #if they do have the same label, check if their spans overlap
                if not spans_overlap: #if they dont overlap in spans, then ignore
                    continue
                if (gold_start, gold_end) == (predicted_start, predicted_end): #and we ignore exact matches
                    continue
                results.append({"sent_idx": sentence_index, "label": gold_label, "gold_span": " ".join(tokens[gold_start:gold_end]), "pred_span": " ".join(tokens[predicted_start:predicted_end]), "gold_idx": (gold_label, gold_end), "pred_idx": (predicted_label, predicted_end), "context": " ".join(tokens[max(0, gold_start - 4):gold_end + 4])})
    return results # save the info of the mismatch


def show_boundary_mismatches(test_sents, test_labels, pred_labels, labels=None, max_per_label=10):
    """this prints boundary mismatches"""
    mismatches = find_boundary_mismatches(test_sents, test_labels, pred_labels) #first we find the mismatches
    mismatches_by_label = defaultdict(list) #create a dictionary to group mismatches by label
    for mismatch in mismatches: #for every mismatch
        label = mismatch["label"] 
        mismatches_by_label[label].append(mismatch) #add the mismatch to its respective label group
    print(
        "Total boundary mismatches "
        "(same label, overlapping, non-exact): "
        f"{len(mismatches)}\n"
    )
    sorted_labels = sorted(mismatches_by_label, key=lambda label: len(mismatches_by_label[label]), reverse=True) #then we sort to see the most common mismatches first

    for label in sorted_labels: #going through labels from most to least common,
        examples = mismatches_by_label[label] #get examples
        print("=" * 70)
        print(f"{label}: {len(examples)} boundary mismatches")
        print("=" * 70)
        for mismatch in examples[:max_per_label]:
            print(f"[sent {mismatch['sent_idx']}]")
            print(f"gold: '{mismatch['gold_span']}' ")
            print(f"{mismatch['gold_idx']}")
            print(f"pred: '{mismatch['pred_span']}' ")
            print(f"{mismatch['pred_idx']}")
            print(f"ctx: ...{mismatch['context']}...")

        print()
    return mismatches_by_label

by_label = show_boundary_mismatches(test_sents, test_labels, pred_labels, labels, max_per_label=10)
```
