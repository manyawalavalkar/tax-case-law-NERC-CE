```python
def remove_spaces_and_labels(input_file, output_file):
    count = 0
    in_sentences = False
    false_o_label = [' \tO\n','\xa0\tO\n','\xa0\tI-PERSON\n', '\xa0\tO\n', '\xa0 \tO\n', '\xa0\xa0\xa0\xa0\xa0\xa0\xa0\xa0\tO\n', ' \tO\n', '\xa0\tI-DATE\n', '\xa0\tI-PROVISION\n', '\xa0\xa0\xa0\xa0\xa0\xa0\tO\n']
    with open(input_file, "r", encoding= "utf-8") as infile, open(output_file, "w", encoding = "utf-8") as outfile:
        for line in infile.readlines():
            # print(repr(line))
            if line in false_o_label:
                continue
            else:
                outfile.write(line)
            if line.strip() == "":
                if in_sentences:
                    count += 1
                    in_sentences = False
            else:
                in_sentences = True
    if in_sentences:
        count += 1
    print(f'number of sentences= {count}')
    return
```

```python
def get_lines(file_path):
    with open(file_path, "r", encoding = "utf-8") as infile:
        all_lines = infile.readlines()
    return all_lines
```
